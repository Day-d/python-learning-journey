#  Smart Text & Word Analyzer Tool

### Project Overview
The **Smart Text & Word Analyzer** is an interactive Python CLI application designed to process, clean, and analyze paragraph-level text inputs. It provides real-time statistical breakdowns, identifies word frequency patterns, highlights word length extremes, and offers an interactive search feature for specific term lookups.

---

### Key Features & Python Concepts Applied
* **Text Processing & Cleaning:** Removes punctuation marks using `string.punctuation` and standardizes text to lowercase to ensure accurate parsing.
* **Basic Analytics:** Calculates total words, character count (with and without spaces), extreme word lengths (longest/shortest), and average word length.
* **Word Frequency & Sorting:** Uses Python **Dictionaries** (`dict`) to count occurrences and applies `sorted()` with `lambda` functions to extract the Top 3 most frequent words.
* **Interactive Search:** Built-in dynamic lookup to find the exact frequency of any specific word entered by the user.
* **Control Flow & State:** Controlled using nested `while` loops, `if-elif-else` conditions, and string manipulations (`strip()`, `lower()`).

---

```python
import string

def analyze_text():
    print("=" * 45)
    print("   SMART TEXT & WORD ANALYZER TOOL")
    print("=" * 45)

    while True:
        text = input("\nEnter a paragraph/text to analyze: ").strip()

        if not text:
            print("There is no text input. Please enter a paragraph.")
            continue

        clean_text = text.translate(str.maketrans('', '', string.punctuation)).lower()
        words = clean_text.split()

        total_words = len(words)
        total_chars_with_spaces = len(text)
        total_chars_no_spaces = len(text.replace(" ", ""))
        longest_word = max(words, key=len) if words else "N/A"
        shortest_word = min(words, key=len) if words else "N/A"
        avg_word_length = total_chars_no_spaces / total_words if total_words > 0 else 0

        word_freq = {}
        for word in words:
            word_freq[word] = word_freq.get(word, 0) + 1

        sorted_freq = sorted(word_freq.items(), key=lambda x: x[1], reverse=True)
        top_words = sorted_freq[:3]

        print("\n" + "-" * 40)
        print("         ANALYSIS RESULTS")
        print("-" * 40)
        print("• Total Words          :", total_words)
        print("• Characters (w/ spaces):", total_chars_with_spaces)
        print("• Characters (no spaces):", total_chars_no_spaces)
        print("• Longest Word         : {} ({} chars)".format(longest_word, len(longest_word)))
        print("• Shortest Word        : {} ({} chars)".format(shortest_word, len(shortest_word)))
        print("• Average Word Length  : {:.2f} characters".format(avg_word_length))

        print("\nTop Frequent Words:")
        for word, count in top_words:
            print("   - '{}' : {} time(s)".format(word, count))

        # 5. Interactive Search Feature (Handles both Words & Symbols like '!')
        print("\n" + "-" * 40)
        search_query = input("Enter a word or symbol to search (or press Enter to skip): ").strip().lower()
        if search_query:
            if search_query in string.punctuation:
                count = text.count(search_query)
                if count > 0:
                    print("Found! The symbol '{}' appears {} time(s).".format(search_query, count))
                else:
                    print("The symbol '{}' was not found in the text.".format(search_query))
            else:
                clean_search = search_query.translate(str.maketrans('', '', string.punctuation))
                count = word_freq.get(clean_search, 0)
                if count > 0:
                    print("Found! The word '{}' appears {} time(s).".format(clean_search, count))
                else:
                    print("The word '{}' was not found in the text.".format(clean_search))

        # 6. Replay Logic
        again = input("\nDo you want to analyze another text or paragraph? (y/n): ").lower().strip()
        if again == 'y':
            continue
        else:
            print("\nThank you for using Text Analyzer! Goodbye.")
            break
```

```python
analyze_text()
```
