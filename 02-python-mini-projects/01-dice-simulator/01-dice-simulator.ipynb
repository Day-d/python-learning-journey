# 🎲 Smart Dice Rolling Simulator

### Project Overview
This interactive Python application simulates dice rolling based on user-defined inputs. It allows users to choose the number of dice sides and total rolls, displays individual outcomes, and automatically calculates basic statistical summaries (Sum and Average).

---

### Key Features & Python Concepts
* **User Input & Validation:** Accepts custom values for rolls and sides using `input()` and integer conversion.
* **Randomization:** Uses Python's `random.randint()` to generate dynamic dice values.
* **Data Storage:** Stores generated outcomes dynamically inside a Python `List`.
* **Built-in Functions:** Computes calculations using `sum()` and `len()`.
* **Control Flow:** Implements a `while` loop for replay logic and `if-elif-else` conditionals for user choices.

---

```python
import random

def roll_dice():
    while True:
        rolls = int(input("How many times do you want to roll?: "))
        sides = int(input("How many sides on the die?: "))

        dice_results = []

        for i in range(rolls):
            result = random.randint(1, sides)
            dice_results.append(result)

        total_sum = sum(dice_results)
        average = total_sum / len(dice_results)

        print("Rolled numbers: {}" .format(dice_results))
        print("Total Sum: {}" .format(total_sum))
        print("Average:{:.2f}".format(average))

        again = input("Do you want to roll again? (y/n/help): ").lower()

        if again == 'help':
            print("Type 'y' to continue or 'n' to exit.")
            again = input("Do you want to roll again? (y/n): ").lower()

        if again == 'y':
            continue

        else:
            print("Thanks for playing! Goodbye")
            break
```

```python
roll_dice()
```
