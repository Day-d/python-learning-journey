# Personal Finance & Expense Tracker Tool

### Project Overview
The **Personal Finance & Expense Tracker** is an interactive Python CLI application designed to record, categorize, and summarize personal incomes and expenses. It provides immediate balance updates and breaks down expenditures by category.

---

### Key Features & Python Concepts Applied
- **Nested Data Structures:** Stores transactions as a `List of Dictionaries` for flexible history tracking.
- **Data Validation & Casting:** Casts inputs to `float` for numerical accuracy and validates dynamic user inputs.
- **Category Summary:** Uses Dictionary aggregation logic to group and total spending by categories (e.g., Food, Transport).
- **Control Flow:** Implements a menu-driven CLI loop using `while` loops, `if-elif-else` branches, and function modularity.

```python
def expense_tracker():
    transactions = []

    print("=" * 45)
    print("PERSONAL FINANCE & EXPENSE TRACKER")
    print("=" * 45)

    while True:
        print("\n--- MENU ---")
        print("1. Add Income")
        print("2. Add Expense")
        print("3. View All Transcations")
        print("4. Summary & Balance")
        print("5. Exit")

        choice = input("Choose an option (1-5): ").strip()

        if choice == '1' or choice =='2':
            trans_type = "Income"  if choice == '1' else "Expense"

            while True:
                try: 
                    amount = float(input(f"Enter {trans_type} Amount: "))
                    if amount <= 0:
                        prin("Amount must be greater than 0.")
                        continue
                    break
                except ValueError:
                    print("Invalid input. Please enter a valid number.")

            category = input("Enter Category (e.g., Salary, Food, Transport): ").strip().capitalize()
            if not category:
                category = "General"

            transactions.append(
                {
                    "type": trans_type,
                    "amount": amount,
                    "category": category
                }
            )
            print(f"Successfully added{trans_type}: ${amount:.2f} [{category}]")

        elif choice == '3':
            print("\n" + "-" * 40)
            print("TRANSACTION HISTORY")
            print("-" * 40)
            if not transactions:
                print("No transactions recorded yet.")
            else:
                for idx, t in enumerate(transactions, 1):
                    print(f"{idx}. [{t['type']}] ${t['amount']:.2f} | Category: {t['category']}") 
            
        elif choice == '4':
            total_income = sum(t['amount'] for t in transactions if t['type'] == 'Income')
            total_expense = sum(t['amount'] for t in transactions if t['type'] == 'Expense')
            net_balance = total_income - total_expense

            category_summary = {}
            for t in transactions:
                if t['type'] == 'Expense':
                    cat = t['category']
                    category_summary[cat] = category_summary.get(cat, 0) + t['amount']

            print("\n" + "-" * 40)
            print("         FINANCIAL SUMMARY")
            print("-" * 40)
            print(f"Total Income  : ${total_income:.2f}")
            print(f"Total Expense : ${total_expense:.2f}")
            print(f"Net Balance   : ${net_balance:.2f}")

            print("\nExpense Breakdown by Category:")
            if not category_summary:
                print("  No expenses recorded yet.")
            else:
                for cat, amt in category_summary.items():
                    print(f"  - {cat}: ${amt:.2f}")

        elif choice == '5':
            print("\nThank you for using Expense Tracker! Goodbye.")
            break
        else:
            print("Invalid choice. Please select between 1 and 5.")
```

```python
expense_tracker()
```
